## Logistic Regression

### (1.1) Logistic Regression

```python
from sklearn.linear_model import LogisticRegression

covariates = [
    "age", "education", "black", "hispanic",
    "married", "nodegree",'re75']

X = nsw[covariates]
T = nsw["treat"]

ps_model = LogisticRegression(max_iter=1000)
ps_model.fit(X, T)

nsw["pscore"] = ps_model.predict_proba(X)[:, 1]
nsw.groupby("treat")["pscore"].describe()
```

### (1.2) Trim Extreme Values

```python
lower = nsw.loc[nsw["treat"] == 1, "pscore"].quantile(0.05) #min()
upper = nsw.loc[nsw["treat"] == 0, "pscore"].quantile(0.95) #max()
nsw_cs = nsw[(nsw["pscore"] >= lower) & (nsw["pscore"] <= upper)]
```

```python
[lower, upper]
```

### (1.3) Nearest-Neighbor for Matching

```python
from sklearn.neighbors import NearestNeighbors
import numpy as np

treated = nsw_cs[nsw_cs["treat"] == 1]
control = nsw_cs[nsw_cs["treat"] == 0]

nn = NearestNeighbors(n_neighbors=1)
nn.fit(control[["pscore"]])

distances, indices = nn.kneighbors(treated[["pscore"]])

matched_controls = control.iloc[indices.flatten()].copy()
matched_treated = treated.copy()

matched_data = pd.concat([matched_treated, matched_controls], ignore_index=True)
```

```python
matched_data.groupby('treat')['re78'].agg(['count','median','mean'])
```

### (1.4) Estimate the Treatment Effect

```python
att = (
    matched_data[matched_data["treat"] == 1]["re78"].mean()
    - matched_data[matched_data["treat"] == 0]["re78"].mean()
)

att
```

### (1.5) Diagnostics I (Standardized Mean Differences)

```python
# Standardized Mean Differences
def smd(df, treat_col, covariate):
    treated_group = df[df[treat_col]==1][covariate]
    control_group = df[df[treat_col]==0][covariate]
    return (treated_group.mean() - control_group.mean()) / np.sqrt((treated_group.var() + control_group.var())/2)

before_smd = [smd(nsw, 'treat', c) for c in covariates]
after_smd = [smd(matched_data, 'treat', c) for c in covariates]

smd_df = pd.DataFrame({
    'Covariate': covariates,
    'Before': before_smd,
    'After': after_smd
}).melt(id_vars='Covariate', var_name='Stage', value_name='SMD')

# Create a single horizontal barplot
plt.figure(figsize=(6, 4))
sns.barplot(data=smd_df, x='SMD', y='Covariate', hue='Stage')
plt.axvline(0, color='black', linestyle='--')
plt.axvline(0.1, color='red', linestyle='--', label='0.1 threshold')
plt.axvline(-0.1, color='red', linestyle='--')
plt.title('Standardized Mean Differences')
plt.xlabel('SMD')
plt.ylabel('Covariates')
plt.legend()
plt.tight_layout()
plt.show()
```

### (1.6) Diagnostics II (Distribution of a Covariate Before and After Matching)

```python
# Plots the distribution of a single covariate before and after matching.
def plot_it(variable):
    fig, axes = plt.subplots(1, 2, figsize=(10, 4))
    # Before Matching
    sns.histplot(data=nsw, x=variable, hue='treat',  stat='density',kde=True,
                 element='step', ax=axes[0])
    axes[0].set_title(f"{variable} – Before Matching")
    axes[0].set_xlabel(variable)
    axes[0].set_ylabel("Density")
    # After Matching
    sns.histplot(data=matched_data, x=variable, hue='treat',  stat='density', kde=True,
                 element='step', ax=axes[1])
    axes[1].set_title(f"{variable} – After Matching")
    axes[1].set_xlabel(variable)
    axes[1].set_ylabel("Density")

    fig.suptitle(f"Distribution of {variable} Before and After Matching", fontsize=16, y=1.05)
    plt.tight_layout()
    plt.show()

# 1) Distribution Plots: Before / After per Variable
for i in covariates:
    plot_it(i)
```

## DoWhy
- 1. Define treatment and outcome
- 2. Create causal model
- 3. Identify effect
- 4.1 Estimate ATE (Average Treatment Effect)
- 4.2 Estimate ATT (Average Treatment effect on Treated)
- 4.3 Estimate CATE (Conditional Average Treatment Effect)


#### (1) - (4)

```python
# 1. Define treatment and outcome
treatment = "treat"
outcome = "re78"

# 2. Create causal model
model = CausalModel(
    data=nsw,
    treatment=treatment,
    outcome=outcome,
    common_causes=["age", "education", "black", "hispanic", "married", "nodegree", "re75"]
)

# 3. Identify effect
identified_estimand = model.identify_effect(proceed_when_unidentifiable=True)

# 4.1 Estimate ATE (Average Treatment Effect)
ate_estimate = model.estimate_effect(
    identified_estimand,
    method_name="backdoor.propensity_score_matching",
    target_units="ate",
    confidence_intervals=True
)
print("Estimated ATE:", ate_estimate.value)
print("95% Confidence Interval (ATE):", ate_estimate.get_confidence_intervals())

# 4.2 Estimate ATT (Average Treatment effect on Treated)
att_estimate = model.estimate_effect(
    identified_estimand,
    method_name="backdoor.propensity_score_matching",
    target_units="att",
    confidence_intervals=True
)
print("Estimated ATT:", att_estimate.value)
print("95% Confidence Interval (ATT):", att_estimate.get_confidence_intervals())

# 4.3 Estimate CATE (Conditional Average Treatment Effect)
# For CATE, we can condition on a subset, e.g., age > 20
subset_data = nsw[nsw["education"] > 8]
cate_model = CausalModel(
    data=subset_data,
    treatment=treatment,
    outcome=outcome,
    common_causes=["age", "education", "black", "hispanic", "married", "nodegree", "re75"]
)
cate_estimand = cate_model.identify_effect(proceed_when_unidentifiable=True)
cate_estimate = cate_model.estimate_effect(
    cate_estimand,
    method_name="backdoor.propensity_score_matching",
    target_units="ate",
    confidence_intervals=True
)
print("Estimated CATE (education > 8):", cate_estimate.value)
print("95% Confidence Interval (CATE):", cate_estimate.get_confidence_intervals())
```

#### (5) Refuters

```python
# --- Random/Common Cause Refuter ---
# Adds a randomly generated covariate to check if the estimate changes spuriously.
ref_random_cc = model.refute_estimate(
    identified_estimand,
    ate_estimate,
    method_name="random_common_cause",
    random_seed=42
)
print("\n[Refuter] Random/Common Cause")
print(ref_random_cc)

# --- Placebo Treatment Refuter ---
# Replaces the real treatment with random noise to see if the "effect" persists spuriously.
ref_placebo = model.refute_estimate(
    identified_estimand,
    ate_estimate,
    method_name="placebo_treatment_refuter",
    placebo_type="permute",     # typical option; permutes treatment
    random_seed=42
)
print("\n[Refuter] Placebo Treatment")
print(ref_placebo)

# --- Data Subset Refuter ---
ref_subset = model.refute_estimate(
    identified_estimand,
    ate_estimate,
    method_name="data_subset_refuter",
    subset_fraction=0.8,         # use 80% of data per draw
    num_simulations=50,          # number of subset re-draws
    random_seed=42
)
print("\n[Refuter] Data Subset")
print(ref_subset)

# --- Bootstrap Refuter ---
ref_bootstrap = model.refute_estimate(
    identified_estimand,
    ate_estimate,
    method_name="bootstrap_refuter",
    num_simulations=200,   # increase for smoother sampling distribution
    sample_size=None,      # None -> same size as original; set int to override
    random_seed=42
)
print("\n[Refuter] Bootstrap")
print(ref_bootstrap)

# --- Add Unobserved Common Cause (simple) ---
# Stress-test sensitivity to hidden confounding using DoWhy defaults.
ref_unobs_cc = model.refute_estimate(
    identified_estimand,
    ate_estimate,
    method_name="add_unobserved_common_cause",
    random_seed=42
)

print("\n[Refuter] Add Unobserved Common Cause")
print(ref_unobs_cc)
```

```python
def run_refuter(model, identified_estimand, estimate, method_name, **kwargs):
    """
    Minimal helper to run a DoWhy refuter and print its result.
    """
    print(f"\n[Refuter] {method_name}")
    result = model.refute_estimate(
        identified_estimand,
        estimate,
        method_name=method_name,
        **kwargs
    )
    print(result)
    return result


# Define all refuters and their (minimal) parameters in one place
REFUTERS = [
    # --- Random/Common Cause Refuter ---
    # Adds a randomly generated covariate to check if the estimate changes spuriously.
    ("random_common_cause", {
        "random_seed": 42
    }),

    # --- Placebo Treatment Refuter ---
    # Replaces the real treatment with random noise to see if the "effect" persists spuriously.
    ("placebo_treatment_refuter", {
        "placebo_type": "permute",   # permute treatment values
        "random_seed": 42
    }),

    # --- Data Subset Refuter ---
    # Re-estimates effect on random subsets to test stability to sample perturbations.
    ("data_subset_refuter", {
        "subset_fraction": 0.8,      # use 80% of data per draw
        "num_simulations": 50,       # number of subset re-draws
        "random_seed": 42
    }),

    # --- Bootstrap Refuter ---
    # Bootstraps the dataset (same size, sampling with replacement) and re-estimates the effect.
    ("bootstrap_refuter", {
        "num_simulations": 200,      # increase for smoother sampling distribution
        "sample_size": None,         # None -> same size as original; set int to override
        "random_seed": 42
    }),

    # --- Dummy Outcome Refuter ---
    # Replaces outcome with random noise; a robust pipeline should yield ~0 effect.
    ("dummy_outcome_refuter", {
        "outcome_type": "noise",     # or "permute"
        "num_simulations": 100,
        "random_seed": 42
    }),

    # --- Add Unobserved Common Cause (simple) ---
    # Stress-test sensitivity to hidden confounding using DoWhy defaults.
    ("add_unobserved_common_cause", {
        "random_seed": 42
        # (Defaults choose direct-simulation and calibrate strengths automatically)
    })
]

# Run all refuters and collect results in a dict
refutation_results = {}
for method_name, params in REFUTERS:
    refutation_results[method_name] = run_refuter(
        model,
        identified_estimand,
        ate_estimate,
        method_name,
        **params
    )
```

### LaLonde DAG

```python
import matplotlib.pyplot as plt
import networkx as nx

# Define nodes
treatment = "treat"
outcome = "re78"
confounders = ["age", "education", "black", "hispanic", "married", "re75"] # "nodegree",

# Create a directed graph
G = nx.DiGraph()

# Add nodes with color attributes
treatment_color = "skyblue"
outcome_color = "lightgreen"
confounder_color = "orange"

G.add_node(treatment, color=treatment_color)
G.add_node(outcome, color=outcome_color)
for conf in confounders:
    G.add_node(conf, color=confounder_color)

# Add edges from confounders to treatment and outcome
for conf in confounders:
    G.add_edge(conf, treatment)
    G.add_edge(conf, outcome)

# Add edge from treatment to outcome
G.add_edge(treatment, outcome)

# Get node colors
colors = [G.nodes[node]['color'] for node in G.nodes]

# Draw the graph
pos = nx.spring_layout(G, seed=42)
plt.figure(figsize=(7, 4))
nx.draw(G, pos, with_labels=True, node_color=colors, node_size=2000, font_size=10, arrows=True)
plt.title("DAG for LaLonde's Dataset", fontsize=16)
```
